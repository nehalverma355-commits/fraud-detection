```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df.head()
```

```python
df_model= df.drop(["nameOrig","nameDest","isFlaggedFraud"], axis =1)
```

```python
df_model.head()
```

```python
categorical =["type"]
numeric =["amount","oldbalanceOrg","newbalanceOrig","oldbalanceDest","newbalanceDest","errorBalanceOrig", "errorBalanceDest", "hourOfDay"]
```

```python
y = df_model["isFraud"]
X = df_model.drop("isFraud",axis=1)
```

```python
X_train,X_test,y_train,y_test =train_test_split(X,y, test_size=0.3 ,stratify=y)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder= "drop"
)
```

```python
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE

smote_pipeline = ImbPipeline([
    ("prep", preprocessor),
    ("smote", SMOTE(sampling_strategy=0.1, random_state=42)), 
    ("clf", LogisticRegression(max_iter=1000))
])

smote_pipeline.fit(X_train, y_train)
y_pred_smote = smote_pipeline.predict(X_test)

print("=== SMOTE results ===")
print(classification_report(y_test, y_pred_smote))
print(confusion_matrix(y_test, y_pred_smote))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

balanced_pipe = Pipeline([("prep", preprocessor),
                           ("clf", LogisticRegression(class_weight="balanced", max_iter=1000))])

scores_balanced = cross_val_score(balanced_pipe, X_train, y_train, cv=cv, scoring="average_precision")
scores_smote = cross_val_score(smote_pipeline, X_train, y_train, cv=cv, scoring="average_precision")

print("class_weight=balanced:", scores_balanced.mean(), "+/-", scores_balanced.std())
print("SMOTE:", scores_smote.mean(), "+/-", scores_smote.std())
```

```python
pipeline= Pipeline([
    ("prep", preprocessor),
    ("clf",LogisticRegression(class_weight="balanced",max_iter=1000))
])
```

```python
pipeline.fit(X_train,y_train)
```

```python
from sklearn.metrics import average_precision_score, roc_auc_score, precision_recall_curve

y_proba = pipeline.predict_proba(X_test)[:, 1]

print("PR-AUC:", average_precision_score(y_test, y_proba))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
```

```python
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


models = {
    "logreg": LogisticRegression(class_weight="balanced", max_iter=1000),
    "random_forest": RandomForestClassifier(
        n_estimators=200, class_weight="balanced", n_jobs=4, random_state=42
    ),
    "xgboost": XGBClassifier(
    n_estimators=300, scale_pos_weight=(y_train==0).sum()/(y_train==1).sum(),
    eval_metric="aucpr", random_state=42, n_jobs=4
    ),
}

results = {}
for name, clf in models.items():
    pipe = Pipeline([("prep", preprocessor), ("clf", clf)])
    pipe.fit(X_train, y_train)
    proba = pipe.predict_proba(X_test)[:, 1]
    results[name] = {
        "pr_auc": average_precision_score(y_test, proba),
        "roc_auc": roc_auc_score(y_test, proba),
    }
import pandas as pd
pd.DataFrame(results).T
```

```python
best_pipe = Pipeline([("prep", preprocessor), ("clf", models["xgboost"])])
best_pipe.fit(X_train, y_train)
y_proba_best = best_pipe.predict_proba(X_test)[:, 1]

precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba_best)

import numpy as np
idx = np.argmax(recalls >= 0.90)
print(f"threshold={thresholds[idx]:.3f}  precision={precisions[idx]:.3f}  recall={recalls[idx]:.3f}")
```
